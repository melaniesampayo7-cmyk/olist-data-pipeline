```python
import pandas as pd
from pathlib import Path

RAW = Path("../raw")
tablas = {}
for archivo in sorted(RAW.glob("*.csv")):
    tablas[archivo.stem] = pd.read_csv(archivo)

for nombre, df in tablas.items():
    print(f"{nombre}: {df.shape[0]} filas, {df.shape[1]} columnas")
```

```python
for nombre, df in tablas.items():
    print("=" * 60)
    print(nombre)
    print("Tipos de datos:")
    print(df.dtypes)
    print("Nulos por columna (solo las que tienen):")
    nulos = df.isna().sum()
    print(nulos[nulos > 0] if (nulos > 0).any() else "ninguno")
    print("Filas duplicadas completas:", df.duplicated().sum())
```

```python
resumen = []
for nombre, df in tablas.items():
    for col in df.columns:
        n_nulos = df[col].isna().sum()
        resumen.append({
            "tabla": nombre,
            "columna": col,
            "tipo": str(df[col].dtype),
            "nulos": n_nulos,
            "% nulos": round(n_nulos / len(df) * 100, 2),
        })
resumen = pd.DataFrame(resumen)

# Solo las columnas que tienen nulos
resumen[resumen["nulos"] > 0]
```

```python
print("customer_id únicos:", tablas["olist_customers_dataset"]["customer_id"].nunique())
print("customer_unique_id únicos:", tablas["olist_customers_dataset"]["customer_unique_id"].nunique())
print("order_id duplicados en orders:", tablas["olist_orders_dataset"]["order_id"].duplicated().sum())
print("review_id duplicados en reviews:", tablas["olist_order_reviews_dataset"]["review_id"].duplicated().sum())
```

```python
orders = tablas["olist_orders_dataset"]
reviews = tablas["olist_order_reviews_dataset"]

# 1) ¿Los nulos de entrega coinciden con pedidos no entregados?
print("Nulos en delivered_customer_date según estado del pedido:")
print(orders[orders["order_delivered_customer_date"].isna()]["order_status"].value_counts())

# 2) Los review_id repetidos, ¿son filas iguales o distintas?
dup = reviews[reviews["review_id"].duplicated(keep=False)].sort_values("review_id")
print("\nEjemplo de review_id repetidos:")
print(dup[["review_id", "order_id", "review_score", "review_creation_date"]].head(10))

# 3) ¿La combinación (review_id, order_id) es única?
print("\nDuplicados por (review_id, order_id):",
      reviews.duplicated(subset=["review_id", "order_id"]).sum())
```

```python
orders = tablas["olist_orders_dataset"]
reviews = tablas["olist_order_reviews_dataset"]
products = tablas["olist_products_dataset"]
translation = tablas["product_category_name_translation"]

# 1) Nulos de aprobación y despacho según estado
print("Nulos en order_approved_at según estado:")
print(orders[orders["order_approved_at"].isna()]["order_status"].value_counts())
print("\nNulos en order_delivered_carrier_date según estado:")
print(orders[orders["order_delivered_carrier_date"].isna()]["order_status"].value_counts())

# 2) Pedidos con más de una reseña
print("\nPedidos con más de una reseña:", reviews["order_id"].duplicated().sum())

# 3) Categorías sin traducción
sin_traduccion = set(products["product_category_name"].dropna()) - set(translation["product_category_name"])
print("Categorías sin traducción:", sin_traduccion)
```
